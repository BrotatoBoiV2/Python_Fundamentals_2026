# Functions

This notebook will teach you how to create and use functions, so that you can
make reusable code blocks which render your code much easier to read and maintain.

By the end of this notebook, you will know how to create a function, call it 
while learning about parameters and arguments.

Functions are created with the `def` keyword followed by the name, parentheses
and a colon to end the starting statement.

## Basic Creation

This is just a basic function creation and calling:

```python
def say_hello():
    print("Hello!")

say_hello() # Output: Hello!
```

## Parameters & Arguments

You can add keywords between the parentheses in the function creation.
Those are parameters for the function, variables that will be used, containing
information passed from an outer scope. (Outside of the function itself)
Parameters can be assigned a default value, making them optional.

When calling the function, you pass it values for the parameters. These are called arguments.
They are set in the order of the parameters, unless explicitly declared.

```python
def output_string(msg):
    print(msg)

output_string("Hola!") # Output: Hola!
output_string("Bonjour!") # Output: Bonjour!
output_string("Kon'nichiwa!") # Output: Kon'nichiwa!

def character(name, player_class, race="Human"):
   print(f"Hello, {name}! You are a {race}, who plays the role of a {player_class}!")

character("Billy", "Ninja") # Output: Hello, Billy! You are a Human, who plays the role of a Ninja!
character("Gruug", "Tinkerer", "Goblin") # Output: Hello, Gruug! You are a Goblin, who plays the role of a Tinkerer!
character(race="Zognoid", player_class="Healer", name="Bob") # Output: Hello, Bob! You are a Zognoid, who plays the role of a Healer!
```

## Extra Features

There are two other features I want to cover that will let you use functions with more control and ease.

These are the `return` keyword, which lets you capture a value from a function, and two advanced parameter types that offer more flexibility.

The extra parameters also introduce a new concept as well. This is called tuple packing `*args` and dictionary packing `**kwargs`.

```python
# Return Statement
def get_sum(num1, num2):
    return num1 + num2

got_sum = get_sum(3, 5)
print(got_sum) # Output: 8

# Advanced Parameters
def shop(*items, **owner):
   print(items) 
   print(owner)

"""
Output:
('Potato', 'Cotton Candy')
{'name': 'BrotatoBoi', 'role': 'Software Developer'}
"""
shop("Potato", "Cotton Candy", name="BrotatoBoi", role="Software Developer")
```
