# Number Guessing Game

This challenge will require you to create a random number guessing game.

## Instructions

The program will generate a number between a given range and allows the user to try 3-6 times - based
on the difficulty - to guess the number correctly.

The program takes these inputs from the user:
* **difficulty:** Allow the user to select the following; easy (6 tries) medium (5 tries) and hard (3 tries)
* **range_start:** What number to start the range at.
* **range_end:** The number to end the range at.
* **guess:** The user's guess.
