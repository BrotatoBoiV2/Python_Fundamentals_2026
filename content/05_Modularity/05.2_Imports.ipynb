# Importing Modules

You will learn about using extra files to organize your code in logical groups and have the code-base be thousands of times easier to read and maintain.

This notebook will teach you how to import an external - or built-in - module, specific parts of a module, and how to install third-party ones.

## Basic Imports

A simple import is using a simple `import` keyword followed by the name of the module you need.
There are many interesting and useful built-in modules you can use in Python, and I highly encourage researching them and learning about them through documentation and experimentation.

````python
import math # Import the entire module
print(math.sqrt(16)) # Output: 4.0


# Specific imports
from random import randint
print(randint(1, 10)) # Output: Random number from 1 to 10.

# Import all items instead of the script, or specific parts.
# However, if the module has a clashing namespace, it will cause silent errors.
from random import *
print(randrange(0, 100, 10)) # Output: Random number from the range 0-100 in steps of 10.

from datetime import datetime as dt # Rename the import for easier calling.
print(dt.now()) # Output: The datetime timestamp of when this line was executed.
````

## Third-Party Modules

You can install other libraries/modules that the Python community has made for the public to use.
This is achieved by using `python-pip` to install the dependencies. For clean installations you can
use Python's built-in module of `venv` to create a virtual environment where you can install all the
packages needed without it being installed on your machine directly.

If you are on Linux you can install it with the following:

### Arch
```bash
sudo pacman -S python python-pip
```

### Ubuntu
```bash
sudo apt-get install python3 python3-pip
```

Then you can create a virtual environment and enter it like so:
```bash
python -m venv .name_of_env
source .name_of_env/bin/activate
```
After that, you are free to install all the dependencies you need.
Importing these modules work the same way as a built-in or custom module.

## Final Tip

You can use an `if` statement to check if the file is the main one or not, so that certain code can only be ran if the file is the main one.

You can do that by doing the following:

````python
if __name__ == '__main__':
    main_func()
````
