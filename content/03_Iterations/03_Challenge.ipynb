# Pyramid Generator

This challenge is to print a pyramid with a custom size and symbol to visually represent it.

## Instructions

The program obtains two forms of input:

* **Pyramid Size:** How many rows should the pyramid be displayed with?
* **Pyramid Symbol:** What symbol should represent the pyramid.

Then it should output the pyramid with the right size and expected symbol.

**Bonus:** Giving the values default options if the user inputs nothing.
