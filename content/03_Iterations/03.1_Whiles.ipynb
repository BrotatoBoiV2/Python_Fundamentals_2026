# While Loops

This loop has a simple statement setup with just its keyword and a boolean statement. However, it adds a bit more complexity since you have to manually increase the iterations and exit the loop yourself. If this is not met, your loop will never escape and the code after it will never execute.

**Note:** If this endless loop is never stopped, it can take up system resources and cause issues.

---

## Basic Loop

There is not much to this besides pairing the keyword `while` with a boolean statement.

```python
"""
Output:
0
1
2
3
4
"""
count = 0
while count != 5:
    print(count)
    count += 1      # Note: `+=` is a shorthand for `count = count + 1`
```

## Precautions

Always ensure that the boolean statement can be switched from inside the loop or it will never end.

### Warning!
**Important Note:** Do not let this cell run too long or it will cause the aforementioned issue. You can kill it by pressing the stop square at the top in the toolbar.

**Script Killswitch:** If you are running this - or a similar - script directly, you can kill it with the key combo of **CTRL**+**C** to send a `KeyboardInterrupt` which kills the program.

```python
"""
Output:
running
running
running
...
.
.
"""
running = True
while running:
    print("running")
    # Always flip the bool from within.
    # running = False

print("this will never be printed")
```
