# For Loops

As a `for` loop statement has more fragments to it, this makes your code much simpler than using a `while` loop that forces you to iterate and exit manually instead of when the value runs dry.
This notebook will teach you how to write and use for loops so that you can iterate through your code and have more control over the script.

By the end of this, you will understand how to write basic for loops, and even some key functions that make for loops a bit more robust.

---

## Basic For Loops

With basic for loops you just create a variable that represents each element within the given value depending on the iteration number.
As the same with the `if` statements, you need to end the `for` statement with a colon `:` and start the next line with an indentation to create the code block.

```python
"""
Output:
H
e
l
l
o
"""
for char in "Hello":
    print(char)
```

## Useful Functions

Instead of giving the for loop a string or a collection (next phase), you can pass a built-in generator function that will give more features to the loop.

The following are a few useful functions to use:
* `range()`: Returns a range of numbers between two values.
* `enumerate()`: Returns the index number of the current element.
* `zip()`: Pairs two values together so they can both be iterated through together.
* There are a few more other generator functions, but they are more advanced.
* You can also create your own. (Phase 5)

### Range

This function allows for three arguments, a start value, an end value and an amount to step by.
Only one is required for the function which is the end value. If only one value is given, it
will start at 0 and end at the provided number. Normally, it will increment by one, but you
can set the step to any number that is needed. You can also count down if you swap the end
and start values (start high end low) and set the step to a negative number.

**Note:** The `range()` function always stops just before the end value. So if `end=10` then it will stop at `9`.

```python
"""
Output:
0
1
2
"""
for i in range(3):
    print(i)

"""
Output:
5
6
"""
for i in range(5, 7):
    print(i)

"""
Output:
0
2
4
6
8
"""
for i in range(0, 10, 2):
    print(i)

"""
Output:
5
4
3
2
1
"""
for i in range(5, 0, -1):
    print(i)
```

### Enumerate

This is a useful function for keeping track of what the index of the current element is. You can also control what number the index starts at.

```python
"""
Output:
0 H
1 e
2 l
3 l
4 o
"""
for index, char in enumerate("Hello"):
    print(index, char)

"""
Output:
10 H
11 e
12 l
13 l
14 o
"""
for index, char in enumerate("Hello", start=10):
    print(index, char)
```

### Zip

This one takes two elements and outputs it the same way the `enumerate()` function does. If both of the elements are of differing sizes, it will go as far as the shorter one. You can bypass that restriction with an external built-in function `zip_longest()` from the module `itertools`. (Phase 5 covers what this means)

```python
"""
Output:
5 H
6 e
7 l
8 l
9 o
"""
for range_num, char in zip(range(5, 10), "Hello"):
    print(range_num, char)

"""
l l
o o
n n
g g
"""
for left_char, right_char in zip("long", "longer"):
    print(left_char, right_char)
```
