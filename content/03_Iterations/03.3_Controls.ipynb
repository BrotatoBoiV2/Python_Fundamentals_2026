# Control Statements

Now that you learned the basics about the loops, this section will teach you control statements. There are three control statements that you can use that have differing and useful functions.

* `break`: Kill a loop.
* `continue`: Continue to next iteration.
* `pass`: Placeholder for empty loops/code blocks.

---

## Break Statement

Use this statement to escape from the loop block entirely. This helps fight against endless loops by overriding the loops statement.

```python
"""
Output:
0
1
2
3
4
"""
count = 0
while True:
    if count == 5:
        break

    print(count)
    count += 1

"""
Output:
0
1
2
3
4
5
"""
count = 0
while True:
    print(count)

    if count == 5:
        break

    count += 1
```

## Continue Statement

The `continue` statement skips the rest of the code within the block and moves on to the next iteration.

```python
"""
Output:
0
1
3
5
6
"""
count = -1
while True:
    count += 1
    if count == 2 or count == 4:
        continue

    print(count)

    if count == 6:
        break
```

## Pass Statement

The `pass` statement is used as a placeholder statement that fills in for an empty code block. Since leaving the statement block empty will throw an `IndentationError`, this control statement helps avoid that.

```python
while True:
    pass
    break

if 1 == 1:
    pass
```

## Final Words

Much like the `if` statements, you can have nested loops within other ones.

**Note:** Nested loops add Time Complexity, which makes the program slower. It also makes it harder for a human to debug it.

```python
while True:
    for number in range(0, 3):
        while 1 == 1:
            break
    break
```

## Quick Mention

You can use an `else` statement at the end of a loop. This works differently than if it was at the end of an `if` statement.
With this `else` statement, it only runs if the loop completed successfully. As in, the loop must not have encountered the `break` statement.

```python
# Output: this runs!
for i in range(10):
    pass
else:
    print("this runs!")

# Output: breaking
for i in range(10):
    if i == 3:
        print("breaking")
        break
else:
    print("this does not run.")
```
