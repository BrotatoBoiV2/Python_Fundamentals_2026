# Error Handling

This section will walk you through how to ensure your scripts are error-proof and won't break during production at 3 A.M. on a holiday.

You should always "sanitize" the user's data instead of blindly trusting it. Blindly trusting the client can always lead to a disaster waiting to happen.

In addition to ensuring the user input is safe to process, you can also wrap fragile/volatile functions with `try`/`except` statements.
That is what this notebook will teach you, how to make your programs more robust.

## Basic Error Handling

Here is a basic example on how to handle errors so your program does not crash.

```python
def error_function():
    items = ["one"]
    print(items[9001]) # Will throw an IndexError since there
    #                      are not that many elements.


try:
    error_function()

# NOTE: Exception is for all errors, but you can
#  replace it with a specific error name. Such as IndexError.
except Exception as e:
    print(e) # This will print the error message that occurred.
```

## Throwing Errors

You can manually throw an error by using the `raise` keyword. This allows you to break out of a function or any chain of logic
much like a `break` statement in a loop, except meaning something went wrong with the logic.

```python
def throw_error():
    user_input = "malformed_data"

    if user_input == "valid":
        return user_input

    else:
        raise ValueError("Received data is malformed and cannot be processed.")

try:
    throw_error()

except ValueError as ve:
    print(f"There is an issue with the data: {ve}") # Always try and make your error messages as
    #                                  detailed, but not too much detail as to overwhelm the user.
```

## Custom Errors

You can even create your own errors so that your codebase can be more descriptive of what is happening.

This will touch on a brand new topic that will be further explored in a future "Python Intermediate" course if this one does well. 
They are classes, a way to create objects and handle their data. We won't get too much into it right now, since that is for a later course.

A custom error is created like so:

```python
# This creates a template class for the error that inherits the methods from the `Exception` class.
# More on this on the next course!
class NoPotatos(Exception):
    pass


spud_amount = 0


def plant_potatos(potatos):
    if potatos == 0:
        raise NoPotatos("There are no potatos to plant!")


try:
    plant_potatos(spud_amount)
except NoPotatos as potatoless:
    print(potatoless)
```
