# Operators

This module will teach you all about the different operator types, how they work and when the best time to use them would be.

This section goes over the four different types of operators that exist and can be used in Python.
The operators fit in the categories of: Arithmetic, Comparison, Logical and Membership.

## Arithmetic

The first type of operators - and the most - that will be covered will be the Arithmetic Operators.
These are the symbols that are used in math and number computations.

When calculating the value, Python follows the PEMDAS rule.
(Parentheses, Exponents, Division/Multiplication, Addition/Subtraction)


### Addition/Subtraction

The addition (`+`) and subtraction (`-`) symbol adds or removes the right value from the left value.
With integers, it works as expected - simple math. When using strings, you can only add (concatenate) 
a string to another one. To remove content from a string you would need a function for string manipulation.

You can find the following examples on how to use these operators.

```python
# Integer Addition.
print(5 + 3) # Output: 8

# String Addition.
print("Hello " + "World!") # Output: Hello World!


# Subtraction.
print(5 - 3) # Output: 2
```

### Multiplication/Division

Similar with the addition/subtraction operators, these operators perform a similar function but with its respective symbol.
Multiplication (`*`) multiplies two integers, however it can also multiply a string - a repeat shorthand. The division symbol (`/`)
divides two numbers as a bit more advanced math.

Here are the examples:

```python
# Integer Multiplication.
print(3 * 3) # Output: 9

# String Multiplication.
print("Hello" * 3) # Output: HelloHelloHello


# Division.
print(6 / 3) # Output: 2.0
```

### Floor, Modulus & Exponents

There are three more mathematical operators that are useful to use sort of like upgraded/special multiplication and division.
Here are the following operators:

**Floor (`//`):** Divides two numbers then rounds them down to the nearest whole number.
***Note:*** For negative numbers it rounds towards negative infinity to the nearest whole number.
**Modulus (`%`):** Divides the two numbers then returns the remainder.
**Exponents (`**`):** Raise a value by the power of another one.

Here are the examples:

```python
# Flooring. (Math - not the carpentry kind)
print(10 // 3) # Output: 3 (instead of 3.333...)

# Modulus.
print(500 % 100) # Output: 0 (100 potatos fit perfectly in 500)
print(317 % 22) # Output: 9 (There are 9 potatos left over)

# Exponentials.
print(4 ** 3) # Output: 64
```

### Multiple Arithmetics

Using that PEMDAS rule, you can calculate any formula in a quick simple line.

#### Breakdown
Original: 4 * 2 + (22 // 2) - (4 ** 2)
First: 4 * 2 + 11 - 16
Second: 8 + 11 - 16
Third: 3

Take this calculation for an example:

```python
# Multi-Operator Calculation
print(4 * 2 + (22 // 2) - (4 ** 2)) # Output: 3
```

## Comparison Operators

These operators are used to compare if two values are equal or not.
As well as check if a number is greater or less than another one.

### Equals/Not-Equal

You can verify if two values are equal to each other or not by using `==` for "is equal to," and `!=` for "is **NOT** equal to."
This applies to any value whether it is a string, integer, float, boolean or any other type that the value might be.

**Note:** It is important to point out that there is a huge difference between a single `=` (Assignment) versus a double `==` (Equality)
which will affect how the script runs, so make sure to use the correct one for the required task.

```python
# Comparing with Equals
print(1 == 1) # Output: True
print(1 == 0) # Output: False
print("Hello" == "Hello") # Output: True
print("Hello" == "hello") # Output: False

# Comparing with Not-Equals
print(1 != 1) # Output: False
print(1 != 0) # Output: True
print("Hello" != "Hello") # Output: False
print("Hello" != "hello") # Output: True
```

### Greater/Less Than (and Equal To)

These operators are used to check if a value is greater/less than another value. With strings, it will check each
character of the string sequentially until it returns a difference, otherwise it returns False.

Additionally, you can also add a `=` after the character to include the right value instead of excluding it.

```python
# Greater Than + Equal To.
print(3 > 2) # Output: True
print(3 > 4) # Output: False
print(3 > 3) # Output: False
print("B" > "C") # Output: False
print("B" > "A") # Output: True
print("A" > "A") # Output: False
print(3 >= 2) # Output: True
print(3 >= 4) # Output: False
print(3 >= 3) # Output: True
print("B" >= "C") # Output: False
print("B" >= "A") # Output: True
print("A" >= "A") # Output: True

# Less Than + Equal To.
print(3 < 2) # Output: False
print(3 < 4) # Output: True
print(3 < 3) # Output: False
print("B" < "C") # Output: True
print("B" < "A") # Output: False
print("A" < "A") # Output: False
print(3 <= 2) # Output: False
print(3 <= 4) # Output: True
print(3 <= 3) # Output: True
print("B" <= "C") # Output: True
print("B" <= "A") # Output: False
print("A" <= "A") # Output: True
```

## Logical Operators

These operators act as logic gates for the code to allow for extra logic flow.
There are three operators that can be used in this section and they are pretty straight-forward.

* `and`: Returns true when only both left and right conditions are True.
* `or`: Returns True when either the left or right condition is True.
* `not`: Flips the boolean of the condition.

Like any other operator, they can be chained together.

```python
# AND Operator
print(True and True) # Output: True
print(True and False) # Output: False
print(False and True) # Output: False
print(False and False) # Output: False

# OR Operator
print(True or True) # Output: True
print(True or False) # Output: True
print(False or True) # Output: True
print(False or False) # Output: False

# NOT Operator
print(not True) # Output: False
print(not False) # Output: True
```

### Logical Notes

You can control the order of logical checks. Much like PEMDAS, you can wrap a condition in parentheses to
control which conditions get checked first in order.

Python takes priority based on the keyword. The following is the order:
   `not`
   `and`
   `or`

You can change this priority with parentheses.

```python
print(not True or False and False) # Output: False
print(not (True or False) and False) # Output: False
print(not ((True or False) and False)) # Output: True
```

#### Breakdown
Condition One: `not True or False and False`
First: `False or False and False`
Second: `False or False`
Final: `False`

Condition Two: `not (True or False) and False`
First: `not True and False`
Second: `False and False`
Final: `False`

Condition Three: `not ((True or False) and False)`
First: `not (True and False)`
Second: `not False`
Final: `True`

## Membership Operators

The fourth and final operators are the Membership Operators. These allow you to check if a value is **in** another one.
For an example, it is used for checking if a string is in a string, or an item in a container. (Future Phase)

There are two different operators with this one; `in` and `not in`

* `in`: Returns True if a value is found within another one.
* `not in`: Mixes the functionality of `in` and `not`. Returns True if the value is not in the other value.

**Note:** You cannot use this with anything besides a container or anything else that is iterable like a string.

```python
print("Hell" in "Hello") # Output: True
print("Hella" in "Hello") # Output: False
print("hello" in "Hello") # Output: False
print("Hell" not in "Hello") # Output: False
print("Hella" not in "Hello") # Output:True
print("hello" not in "Hello") # Output: True
```
