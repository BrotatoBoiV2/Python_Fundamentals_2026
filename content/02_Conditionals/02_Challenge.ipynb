# Login Menu

This challenge is to make a script that allows the user access to "secret information" that can only - in theory - be accessed by the only one with the right credentials. Anything extra is encouraged, but as long as it can function as a login for at least a single user.

## Instructions

The script must ask the user for their username and password, then compare if it is correct.

If the username exists and the password is correct, allow them access. Otherwise, tell them something went wrong and shut down the program.
Never specify why though, the ambiguity helps against enumeration (brute-force) attacks.
