your scripts will have logic flow and be capable of making "decisions."

## If/Else

The first type of conditional statements one learns in Python is a chain of `if`/`elif`/`else` which
allows to add logical decisions within your script.

* `if`: The first branch. Every statement must start with it.
* `elif`: The next branch that gets checked if the previous branch does not run.
* `else`: If this final branch is added, it always runs as long as none of the previous branches ran.

### Important Note

Each branch statement needs to end with a colon `:` and the code the branch needs to run has to be indented.
Indentation is a four spaced tab, this has to be consistent throughout. I personally use three spaced tabs,
but when collaborating and learning it is best to stick with the PEP-8 standard of four spaced tabs to avoid any
indentation errors with external code.

Also, each `if` statement will always try to evaluate, so if you need branch b to run if branch a is not True, then you need to use 
an `elif` to link them. Otherwise, both of the `if` statements will run regardless of how previous branches evaluated.

An `else` statement cannot be placed between two `elif` branches for that will raise an error. Keep the `else` at the bottom.

```python
# Output: Hello
if True:
    print("Hello")
elif False:
    print("Hola")
else:
    print("None")

# Output: Hola
if False:
    print("hello")
elif True:
    print("Hola")
else:
    print("None")

# Output: None
if False:
    print("Hello")
elif False:
    print("Hola")
else:
    print("None")   
```

## Match/Case

These statements work much like the `if`/`elif`/`else` statements but are preferable when working with long `if`-`elif` branches - about three branches would warrant switching. It is also useful for checking the shape of complex data, like looking for a list that starts with a keyword. To create this code block, you start by pairing the `match` keyword with the value you want to check, you then end it with a colon `:` and start a new line with an indent. Next, you create a `case` statement with the target value. Much like the `match` statement, the `case` statement ends with a colon `:` and moves onto the next line with an extra indentation. Finally, this indentation is where the code that the `case` statement should run will be written.

**Note:** This feature is only available in Python 3.10 and later.

### Basic Usage

This is the most simple version of a `match`/`case` statement:

```python
# Output: sauce
fruit = "apple"
match fruit:
    case "apple":
        print("sauce")
    case "banana":
        print("split")

# Output: split
fruit = "banana"
match fruit:
    case "apple":
        print("sauce")
    case "banana":
        print("split")
```

### Extra Features

There are four extra features that can be used to make the case statement function more in-depth.

The following features are available to be used:

* An `or` statement can be simulated by using a pipe `|` character between each of the values.
* A wildcard can be created by inserting an underscore `_` instead of a value, this acts as an `else` statement.
* You can use guards on the `case` statement by pairing it with an `if` statement.
* Extracting values from a tuple or something is able to be done as well to process things like positions and such.


#### Pipes & Wildcards

**Note:** Much like the `else` statement, placing the wildcard between two other case statements will cause issues with your script. Keep the wildcard at the bottom.

```python
# Output: fruit
fruit = "apple"
match fruit:
    case "apple" | "banana":
        print("fruit")
    case _:
        print("unknown")

# Output: fruit
fruit = "banana"
match fruit:
    case "apple" | "banana":
        print("fruit")
    case _:
        print("unknown")

# Output: unknown
fruit = "uranium"
match fruit:
    case "apple" | "banana":
        print("fruit")
    case _:
        print("unknown")
```

#### Guards

```python
# Output: ready
ripe_level = 95
fruit = "apple"

match fruit:
    case "apple" if ripe_level > 65:
        print("ready")
    case _:
        print("not ready")

# Output: not ready
ripe_level = 60
fruit = "apple"

match fruit:
    case "apple" if ripe_level > 65:
        print("ready")
    case _:
        print("not ready")
```

#### Value Unpacking

**Note:** Whatever variable you unpack (i.e. `x`) will be kept in memory out of scope (the code block) so it will still be
accessible when the `match` statement finishes executing. This can cause issues if the unpacking `case` never ran, causing a `NameError`.

(You can test this by clearing the memory just click **"Restart Kernel & Clear Outputs"**)

```python
# Output: 69 420
#         69
# This script can cause a `NameError`
# if the second branch never ran.
pos = (69, 420)
match pos:
    case (0, 0):
        print("origin")

    case (x, y):
        print(x, y)

print(x)
```

```python
# Output: origin
# This one will not error out
# since the unpacked variable
# is not being called outside.
pos = (0, 0)
match pos:
    case (0, 0):
        print("origin")
    case (x, y):
        print(x, y)
```
