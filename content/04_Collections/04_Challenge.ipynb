# Inventory Management Script

The goal of this challenge is to create an inventory management script for anything you would like.

## Instructions

A simple program that keeps track of an inventory of items and allows for the following functions:

* `add`: Add an item to the inventory.
* `remove`: Remove an item from the inventory.
* `display`: Display the items in the inventory.

Ensure to give feedback when the user tries to delete an item that does not exist or adds an item that already exists.

### Bonus

If you want an extra challenge, you can ***pair*** the items with a **price** and manipulate those values as well.
