# Sets

In this notebook, you will learn about sets and how they can be utilized within your scripts.
These are unordered collection types that only contain unique elements. These elements cannot be changed,
but elements can be added and removed to the set.

There are a handful of methods that help manipulate these sets:

* `add(element)`: Add an element to the set.
* `update(iterable)`: Update the set with items from a list. 
* `remove(element)`: Remove an element from a set. Throws a `KeyError` if the element does not exist.
* `discard(element)`: Remove the element from a set. Does not throw an error, just moves on if element is not found.

Additionally, there are extra methods that do mathematical functions:

* `union(setB)`: Return all unique elements from both of the sets.
* `intersection(setB)`: Return all elements that are in both sets.
* `difference(setB)`: Return the elements found in A that are not in B.
* `symmetric_difference(setB)`: Return elements that are in either sets, but not both.

**Note:** Sets are always unordered and may output differently based on an internal hash calculation for strings. While integers evaluate to their own hashes and might look perfectly sorted in small examples, relying on this behavior for sorting is dangerous. Never write code that assumes or relies on a specific set order.

## Basic Creation

```python
elements = {"Fluorine", "Uranium", "Carbon"}
print(elements) # Output: {'Fluorine', 'Uranium', 'Carbon'}

elements.add("Potassium")
print(elements) # Output: {'Fluorine', 'Uranium', 'Carbon', 'Potassium'}

elements.update(["Yttrium", "Oxygen", "Uranium"])
print(elements) # Output: {'Fluorine', 'Uranium', 'Carbon', 'Potassium', 'Yttrium', 'Oxygen'}

elements.remove("Carbon")
print(elements) # Output: {'Fluorine', 'Uranium', 'Potassium', 'Yttrium', 'Oxygen'}

elements.discard("Carbon") # Note: If `.remove()` was used, a `KeyError` would be thrown.
print(elements) # Output: {'Fluorine', 'Uranium', 'Potassium', 'Yttrium', 'Oxygen'}
elements.discard("Yttrium")
print(elements) # Output: {'Fluorine', 'Uranium', 'Potassium', 'Oxygen'}
```

## Mathematical Operators

These methods work in a different way, and they have their own operator that makes using it easier.
For these example methods, I will be using the respective operators. Each one is called as: `setA.method(setB)`.

Here are the operators for their respective methods:

|        Method        | Operator |
| -------------------- | -------- |
|         union        |    \|    |
|     intersection     |    &     |
|      difference      |    -     |
| symmetric_difference |    ^     |

```python
first_set = {1, 2, 3, 4, 5, 6}
second_set = {3, 4, 5, 6, 7, 8, 9}
third_set = {6, 7, 8, 9, 10, 11, 12}

print(first_set | second_set) # Output: {1, 2, 3, 4, 5, 6, 7, 8, 9}
print(second_set & third_set) # Output: {6, 7, 8, 9}
print(third_set - second_set) # Output: {10, 11, 12}
print(first_set ^ third_set) # Output: {1, 2, 3, 4, 5, 7, 8, 9, 10, 11, 12} 
```
