# Dictionaries

You will learn about dictionaries in this notebook and how to use and access their `key:value` pairs.

Dictionaries are a collection type that allows the elements within to be manipulated. Similar to the `Lists` and `Tuples`
objects, the dictionary elements are ordered. Each element is a `key:value` pair where the `key` has to be unique and
must be immutable types such as strings, tuples, numbers, and so on. Values can represent any type of value from strings,
integers and nested dictionaries.

* `keys()`: Returns an object of dict_keys that contain a list of all of the keys. 
* `values()`: Returns an object of dict_values that contain a list of all of the values.
* `items()`: Returns an object of dict_items that contain a list of tuples of the keys and their paired values.
* `update(dict)`: Update the dictionary with new entries.
* `clear()`: Clear the entire dictionary.
* `pop(key)`: Return the value of the key and delete that entry.
* `get(key, default)`: A safe way to obtain a key. Returns a default value if not found.

```python
profile = {
    "name": "Billy",
    "age": 999
}

print(profile.keys()) # Output: dict_keys(['name', 'age'])
print(profile.values()) # Output: dict_values(['Billy', 999])
print(profile.items()) # Output: dict_items([('name', 'Billy'), ('age', 999)])
profile.update({"email": "totallyrealmail123@aol.com"})
print(profile.items()) # Output: dict_items([('name', 'Billy'), ('age', 999), ('email', 'totallyrealmail123@aol.com')])
email = profile.pop("email")
print(email) # Output: totallyrealmail123@aol.com
print(profile) # Output: {'name': 'Billy', 'age': 999}
print(profile.get("email", "Email not found!")) # Output: Email not found!

# Another way to remove an entry:
# NOTE: Using this way is fragile.
#   Ensure the key exists (e.g. `in`)
#   so you do not receive a `KeyError`.
del profile["name"]
print(profile) # Output: {'age': 999}

profile.clear()
print(profile) # Output: {}

# Safe way to use the `del` keyword.
profile.update({"name": "delete"})
if "name" in profile:
    del profile["name"]
```
