# Tuples

These are unchangeable, ordered collection types. They are normally used with the same type of data and only contain 2 functions since they cannot be manipulated.

* `count(element)`: Return the amount of times that element appears.
* `index(element)`: Returns the index of the first occurring element found.

## Basic Creation

You can create a tuple in many different ways:

```python
position = (0, 0)
last_pos = tuple() # Safely create an empty tuple
first_dimension = (21,) # A trailing comma is needed, otherwise it will just be an integer.
single_item = ("poutine", ) # This is the same as the above but would return a string without the comma.
# TL;DR: No trailing comma, no single item tuple.
space_3d = 36, 767, 36 # Does not need explicit parentheses when declared like this.

print(position) # Output: (0, 0)
print(last_pos) # Output: ()
print(first_dimension) # Output: (21, )
print(space_3d) # Output: (36, 767, 36)
```

## Manipulate Tuple

To say it is impossible to "manipulate" the variable would be a lie. You can theoretically change the value by doing something like the following:

```python
last_pos = position
position = (position[0], position[1]+1)

print(last_pos) # Output: (0, 0)
print(position) # Output: (0, 1)
```

## Utility Functions

There are two utility functions that can be executed on the tuples. You can count how many times an element appears in the tuple, and find the index number of the first occurrence of an element.

```python
print(last_pos.count(0)) # Output: 2
print(space_3d.index(36)) # Output: 0
```

## Tuple Unpacking

You can unpack all of the elements into variables very easily by doing the following:

```python
x, y, z = space_3d
print(x, y, z) # Output: 36 767 36

four_dim =  (x, y, z, 420)
print(four_dim) # Output: (36, 767, 36, 420)

x, y, *extra_coords = four_dim
print(x, y, extra_coords) # Output: 36 767 [36, 420]
```
