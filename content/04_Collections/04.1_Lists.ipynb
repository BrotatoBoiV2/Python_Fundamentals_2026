# Lists

Lists are a basic collection of elements which could be of any Python object.
They have some features that will help you manipulate them as much as possible.
They can contain a mixture of different objects and contain duplicates. Elements
are accessed through an indexing-based system, where the first element starts at 0.

There are a handful of useful functions that you can perform on lists to manipulate them:

* `append(element)`: Append an element to the list.
* `insert(index, element)`: Insert an element at the index spot.
* `extend(iterable)`: Adds the contents of an iterable to the list.
* `remove(element)`: Removes the first occurrence of that element.
* `pop(index)`: Removes and returns the element at the index, if not given it removes the last.
* `sort()`: Sort the elements in the list.
* `reverse()`: Reverses the order of the list.
* `count(element)`: Return the amount of times the element is found.

```python
"""
Output:
apples
bananas
oranges
"""
fruits = ["apples", "bananas", "oranges"]
sauces = list() # Creates an empty list, same as `[]`
for fruit in fruits:
    print(fruit)

print(fruits[0]) # Output: apples
print(fruits[2]) # Output: oranges
```

## Element Addition

You can add elements to the list by appending them to the end, inserting them at a
specific index number, or extending the list by adding another iterable element.

```python
sauces.append("bbq")
print(sauces) # Output: ['bbq']

fruits.insert(0, "bananas")
print(fruits) # Output: ['bananas', 'apples', 'bananas', 'oranges']

extra_sauce = ["ketchup", "mustard", "mayo"]
sauces.extend(extra_sauce)
print(sauces) # Output: ['bbq', 'ketchup', 'mustard', 'mayo']
```

## Element Removal

Alternatively, you can also remove the first element found in a list, or pop out an
element from the list so it is no longer in it, but return the value so you can use
it later.

```python
sauces.remove("mustard")
print(sauces) # Output: ['bbq', 'ketchup', 'mayo']

pick_fruit = fruits.pop(1)
print(fruits) # Output: ['bananas', 'bananas', 'oranges']
print(pick_fruit) # Output: apples
pick_sauce = sauces.pop()
print(pick_sauce) # Output: mayo
print(sauces) # Output: ['bbq', 'ketchup']
```

## List Utilities

There are a few useful utility functions that you can use with a list.
These are the sorting function, reversing function, and the element counter.

**Note:** The `.sort()`, `.reverse()` and other methods alike, modify the list directly and do not return a value. Therefore, assigning these to a variable will just yield `None` which would not be useful.

```python
# This also works with strings.
# It just goes lexicographically.
# ASCII/Unicode ordering.
numbers = [5, 6, 2, 7, 10]
numbers.sort()
print(numbers) # Output: [2, 5, 6, 7, 10]
numbers.reverse()
print(numbers) # Output: [10, 7, 6, 5, 2]

banana_count = fruits.count("bananas")
print(banana_count) # Output: 2
```

## Extra Functionalities

There are two extra features of the lists that can really help make your code much better.

There are List Comprehension and List Slicing.

List comprehension is a quick and concise way to create a list from a `for` loop.
List Slicing allows you to select a portion of the list so that you can process a certain window of information.

```python
# List Comprehension
numbers = [5, 2, 68, 9, 6]
plus_one = [num + 1 for num in numbers]
print(plus_one) # Output: [6, 3, 69, 10, 7]

# List Slicing
print(plus_one[1:3]) # Output: [3, 69]
```
